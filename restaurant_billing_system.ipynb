```python
print("--------------Menu----------------")
menu = {
    # Starters
    "Tomato Soup": 120,
    "Sweet Corn Soup": 130,
    "Hot & Sour Soup": 140,
    "Spring Rolls": 150,
    "Garlic Bread": 100,
    "Paneer Tikka": 180,
    "Chicken Tikka": 220,
    "Veg Manchurian Dry": 160,
    "Chilli Paneer": 170,
    "French Fries": 90,

    # Main Course - Indian
    "Paneer Butter Masala": 250,
    "Dal Tadka": 180,
    "Dal Makhani": 200,
    "Chole Bhature": 150,
    "Rajma Rice": 130,
    "Veg Biryani": 220,
    "Chicken Biryani": 300,
    "Mutton Biryani": 350,
    "Fish Curry": 280,
    "Veg Fried Rice": 180,

    # Main Course - Continental
    "Margherita Pizza": 220,
    "Farmhouse Pizza": 280,
    "Cheese Burger": 150,
    "Veg Burger": 120,
    "Grilled Sandwich": 100,
    "Club Sandwich": 140,
    "Pasta Alfredo": 200,
    "Pasta Arrabiata": 190,
    "Lasagna": 250,
    "Veg Sizzler": 300,

    # South Indian
    "Masala Dosa": 140,
    "Plain Dosa": 120,
    "Idli Sambar": 100,
    "Vada Sambar": 110,
    "Uttapam": 130,
    "Rava Dosa": 150,
    "Onion Dosa": 160,
    "Set Dosa": 140,
    "Pesarattu": 150,
    "Medu Vada": 120,

    # Desserts
    "Ice Cream": 100,
    "Gulab Jamun": 70,
    "Rasgulla": 80,
    "Brownie": 120,
    "Cheesecake": 200,
    "Fruit Salad": 90,
    "Kulfi": 60,
    "Falooda": 150,
    "Chocolate Mousse": 180,
    "Apple Pie": 160,

    # Drinks
    "Coffee": 90,
    "Cold Coffee": 100,
    "Lemonade": 70,
    "Soft Drink": 60,
    "Milkshake": 120,
    "Iced Tea": 110,
    "Green Tea": 100,
    "Hot Chocolate": 130,
    "Mocktail": 150,
    "Fresh Juice": 120,

    
    #Other items
    "Paneer Kadhai": 240,
    "Butter Naan": 40,
    "Tandoori Roti": 20,
    "Stuffed Paratha": 80,
    "Veg Pulao": 200,
    "Chicken Curry": 280,
    "Mutton Curry": 350,
    "Fish Fry": 300,
    "Egg Curry": 180,
    "Veg Kofta": 220,
    "Veg Handi": 230,
    "Paneer Bhurji": 210,
    "Veg Hakka Noodles": 190,
    "Chicken Hakka Noodles": 220,
    "Veg Chowmein": 180,
    "Chicken Chowmein": 220,
    "Paneer Chilli": 200,
    "Veg Momos": 120,
    "Chicken Momos": 150,
    "Tandoori Chicken": 320,
}
for item, price in menu.items():
    print(f"{item:<25} : ₹{price}")
try:
    expenses = []  
    
    def add_item():
        item = input("Enter food item: ")
        quantity = int(input("How many products want to add? "))

        if quantity > 0:
            print("Valid quantity")
    
            if item in menu:
                price = quantity * menu[item]
                expenses.append([item, quantity, price])
                print(f"Added {item} x{quantity} = ₹{price}")
            else:
                print("Item not found in menu!")
    
        else:
            print("Please enter a positive number")
       
    
    def view_order():
        print("------ Expenses ------")
        for e in expenses:
            print(f"{e[0]:<12} :{e[1]} :₹{e[2]}")
    def remove_item():
        remove = input("Enter the item you want to remove: ")
        found = False
        for e in expenses:
            if e[0].lower() == remove.lower():  
                expenses.remove(e)
                found = True
                print(f"{remove} (Qty:{e[1]}, ₹{e[2]}) has been removed successfully")
                break   
        if not found:
            print("The item is not in list, please check again")

    
    def calculate_total():
        total = sum(e[2] for e in expenses)
        print("Total expense is:", total)
    def total_bill():
        if expenses:
            subtotal = sum(e[2] for e in expenses)   
            gst_rate = 0.05                          
            gst_amount = subtotal * gst_rate
            total = subtotal + gst_amount
            print("---------Final Bill--------")
            print("Item                           Qty  Price")
            for item, quantity, price in expenses:
                    print(f"{item:<30} {quantity:<3} ₹{price:<5}")
            print("Subtotal:",subtotal)
            print("GST (5%):",gst_amount)
            print("Total Bill:",total)
        else:
            print("List is empty")

        
    def highest_item():
        if expenses: 
            maximum = expenses[0]   
            for e in expenses:      
                if e[2] > maximum[2]:   
                    maximum = e        
            print("Highest expense is:", maximum[0], "₹", maximum[2])
        else:
            print("No expenses yet!")
    
    while True:
        print("1. Add Item")
        print("2. View Order")
        print("3.Remove item")
        print("4. Calculate Total")
        print("5. Highest-Priced Item")
        print("6.Total bill")
        print("7. Exit")
        
        choice=int(input("Enter a choice :"))
        if choice==1:
            add_item()
        elif choice==2:
            view_order()
        elif choice==3:
            remove_item()
        elif choice==4:
            calculate_total()
        elif choice==5:
            highest_item()
        elif choice==6:
            total_bill()
        elif choice == 7:
            print("------Exiting-------")
            if expenses:
                print("Final Expenses:")
                for e in expenses:
                    print(f"{e[0]:<15} Qty:{e[1]}  Total: ₹{e[2]}")  
            else:
                print("The list is empty ")
            break 
        else:
            print("Invalid choice! Please enter between 1 and 7")
except ValueError:
            print("Invalid input!")
```
